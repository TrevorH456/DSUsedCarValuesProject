```python
# Importing the files and begin initial setup
!mamba install pandas
import pandas as pd
cars = pd.read_csv("used_cars.csv")

# Test that loading was successful and working properly
display(cars.head(10))
display(cars.tail(10))

# Check shape and data types
display(cars.shape)
display(cars.dtypes)
cars.info

# This is only just the setup portion for the project, but will be important for the rest of develeopment
```

```python
# Precheck for missing and duplicate entries
display(cars.isnull().sum())
display(cars.duplicated().sum())

# Found some empty values, so going to fix that
```

```python
# Missing some entries (fuel_type, accident, clean_title)
# Going to fix them and fill it in with 'Unknown'
cars["fuel_type"] = cars["fuel_type"].fillna("Unknown")
cars["accident"] = cars["accident"].fillna("Unknown")
cars["clean_title"] = cars["clean_title"].fillna("Unknown")
display(cars.isnull().sum())

# Dataset is no longer missing values, moving on now
```

```python
# Two cols are not numeric from check earlier, but need to be (price and milage)
cars["milage"] = (
    cars["milage"]
    .str.replace(",", "", regex=False)
    .str.replace(" mi.", "", regex=False)
    .astype(int)
)
cars["price"] = (
    cars["price"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(int)
)

# Check to make sure changes went through
display(cars[["milage", "price"]].head())
display(cars[["milage", "price"]].dtypes)
# Check the data type now
cars.describe()

# Dataset now has milage and price as numerical type instead of object, so we can use operations correctly without error next
```

```python
# Run some numerical tests on some of the values to show process working
display(cars[["model_year", "milage", "price"]].describe())

# Couple more tests for fun
# Lowest and Highest vehicle prices
display(cars.nsmallest(10, "price")[["brand", "model", "model_year", "milage", "price"]])
display(cars.nlargest(10, "price")[["brand", "model", "model_year", "milage", "price"]])
# Lowest and Highest vehicle milage
display(cars.nsmallest(10, "milage")[["brand", "model", "model_year", "milage", "price"]])
display(cars.nlargest(10, "milage")[["brand", "model", "model_year", "milage", "price"]])

# Lots of things to take note here
# year range from 1974-2024 // milage from 100-405000 // price from $2000-$44553
# got some exotic cars, so the mean would definitely be higher from it
```

```python
# Make some visuals so to get a better idea of the numerical stats
import matplotlib.pyplot as plt

# Graph for cars under $200k
cars_under_200k = cars[cars["price"] <= 200000]
plt.figure(figsize=(8, 5))
plt.hist(cars_under_200k["price"], bins=30)
plt.title("Variety of Used Car Prices Under $200,000")
plt.xlabel("Price ($)")
plt.ylabel("Frequency")
plt.show()

# Graph of milage variety
plt.figure(figsize=(8, 5))
plt.hist(cars["milage"], bins=20)
plt.title("Vehicle Mileage Variety")
plt.xlabel("Mileage")
plt.ylabel("Frequency")
plt.show()

# Graph for prices by car brand
avg_price_brand = (
    cars.groupby("brand")["price"]
    .mean()
    .sort_values(ascending=False)
    .head(15)
)
plt.figure(figsize=(10, 6))
avg_price_brand.plot(kind="bar")
plt.title("Brands by Average Used Car Price")
plt.xlabel("Brand")
plt.ylabel("Average Price ($)")
plt.xticks(rotation=45, ha="right")
plt.show()

# Graphs help with visualizing how much gap there is within this dataset (old used car VS brand new supercar)
# Might have to see if treating higher-end cars seperately cwould be better for the project result
```

```python
# Starting the machine learning 
# Predictor and Target variable setups
X = cars[
    [
        "brand",
        "model",
        "model_year",
        "milage",
        "fuel_type",
        "accident",
        "clean_title"
    ]
]
y = cars["price"]

# Split data to the training set and tetsting set'
!mamba install scikit-learn
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)
print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)

# Setup categorical vars for test
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
categorical_columns = [
    "brand",
    "model",
    "fuel_type",
    "accident",
    "clean_title"
]
numeric_columns = [
    "model_year",
    "milage"
]
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_columns),
        ("num", "passthrough", numeric_columns)
    ]
)
linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", LinearRegression())
    ]
)

# Training the data
linear_model.fit(X_train, y_train)
# Predict prices with data
linear_predictions = linear_model.predict(X_test)

# Evaluating Test Results
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Running tests on regression earlier
linear_mae = mean_absolute_error(y_test, linear_predictions)
linear_mse = mean_squared_error(y_test, linear_predictions)
linear_rmse = np.sqrt(linear_mse)
linear_r2 = r2_score(y_test, linear_predictions)
print("Linear Regression Results")
print("MAE:", linear_mae)
print("MSE:", linear_mse)
print("RMSE:", linear_rmse)
print("R²:", linear_r2)

# The predicitons are off by $28k, and R2 measures as far away from 1.0 as possible, so the exotic cars are not helping me here
# Probably will have to remove them, but we'll see
```

```python
# Decision Tree Regression Evaluation
# Testing other regression method
from sklearn.tree import DecisionTreeRegressor

# Decision Tree Regression model
decision_tree_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", DecisionTreeRegressor(random_state=42))
    ]
)
# Train the model for data
decision_tree_model.fit(X_train, y_train)
# Make predictions 
tree_predictions = decision_tree_model.predict(X_test)

# Evaluate test results
tree_mae = mean_absolute_error(y_test, tree_predictions)
tree_mse = mean_squared_error(y_test, tree_predictions)
tree_rmse = np.sqrt(tree_mse)
tree_r2 = r2_score(y_test, tree_predictions)
print("Decision Tree Regression Results")
print("MAE:", tree_mae)
print("MSE:", tree_mse)
print("RMSE:", tree_rmse)
print("R²:", tree_r2)

# Only a bit better than Decision Tree, off by 20k this time
# Getting better results, next is Random Forest; hopefully it's better
```

```python
# Random Forest Regression Testing
# Begin the test
from sklearn.ensemble import RandomForestRegressor

# Random Forest Regression model
random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", RandomForestRegressor(
            n_estimators=100,
            random_state=42
        ))
    ]
)
# Train the model for data
random_forest_model.fit(X_train, y_train)
# Make predictions
forest_predictions = random_forest_model.predict(X_test)

# Evaluate test results
forest_mae = mean_absolute_error(y_test, forest_predictions)
forest_mse = mean_squared_error(y_test, forest_predictions)
forest_rmse = np.sqrt(forest_mse)
forest_r2 = r2_score(y_test, forest_predictions)
print("Random Forest Regression Results")
print("MAE:", forest_mae)
print("MSE:", forest_mse)
print("RMSE:", forest_rmse)
print("R²:", forest_r2)

# The model is the best option here, only off by 18k, and R2 is closer to 1.0 than others
# Still not great prediction, but I'll work with what it is
```

```python
# Testing again without exotic car prices in test set
# Setup dataset window (repeat of earlier, just new set)
cars_filtered = cars[cars["price"] <= 200000].copy()

print("Original dataset size:", cars.shape)
print("Filtered dataset size:", cars_filtered.shape)
X_filtered = cars_filtered[
    [
        "brand",
        "model",
        "model_year",
        "milage",
        "fuel_type",
        "accident",
        "clean_title"
    ]
]
y_filtered = cars_filtered["price"]

X_train_f, X_test_f, y_train_f, y_test_f = train_test_split(
    X_filtered,
    y_filtered,
    test_size=0.25,
    random_state=42
)

# Linear regression training
linear_model_f = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", LinearRegression())
    ]
)
linear_model_f.fit(X_train_f, y_train_f)
linear_predictions_f = linear_model_f.predict(X_test_f)
linear_mae_f = mean_absolute_error(y_test_f, linear_predictions_f)
linear_rmse_f = np.sqrt(mean_squared_error(y_test_f, linear_predictions_f))
linear_r2_f = r2_score(y_test_f, linear_predictions_f)

print("Linear Regression Results")
print("MAE:", linear_mae_f)
print("RMSE:", linear_rmse_f)
print("R²:", linear_r2_f)

# Decision Tree regression training
decision_tree_model_f = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", DecisionTreeRegressor(random_state=42))
    ]
)
decision_tree_model_f.fit(X_train_f, y_train_f)
tree_predictions_f = decision_tree_model_f.predict(X_test_f)
tree_mae_f = mean_absolute_error(y_test_f, tree_predictions_f)
tree_rmse_f = np.sqrt(mean_squared_error(y_test_f, tree_predictions_f))
tree_r2_f = r2_score(y_test_f, tree_predictions_f)

print("Decision Tree Regression Results")
print("MAE:", tree_mae_f)
print("RMSE:", tree_rmse_f)
print("R²:", tree_r2_f)

# Random Forest regression training
random_forest_model_f = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", RandomForestRegressor(
            n_estimators=100,
            random_state=42
        ))
    ]
)
random_forest_model_f.fit(X_train_f, y_train_f)
forest_predictions_f = random_forest_model_f.predict(X_test_f)
forest_mae_f = mean_absolute_error(y_test_f, forest_predictions_f)
forest_rmse_f = np.sqrt(mean_squared_error(y_test_f, forest_predictions_f))
forest_r2_f = r2_score(y_test_f, forest_predictions_f)

print("Random Forest Regression Results")
print("MAE:", forest_mae_f)
print("RMSE:", forest_rmse_f)
print("R²:", forest_r2_f)

# Filtering out the exotic cars really helped all three regression test and their training results
# Random Forest did the best again, with $10k off actual price; still a bit off, but way better than it was before
```
